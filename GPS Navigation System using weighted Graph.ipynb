```python
import heapq

class GPSNavigation:
    def __init__(self):
        self.graph = {}

    def add_road(self, city1, city2, distance):
        if city1 not in self.graph:
            self.graph[city1] = []

        if city2 not in self.graph:
            self.graph[city2] = []

        self.graph[city1].append((city2, distance))
        self.graph[city2].append((city1, distance))

    def shortest_path(self, start, destination):
        distances = {city: float('inf') for city in self.graph}
        distances[start] = 0

        previous = {}
        queue = [(0, start)]

        while queue:
            current_distance, current_city = heapq.heappop(queue)

            if current_distance > distances[current_city]:
                continue

            if current_city == destination:
                break

            for neighbor, weight in self.graph[current_city]:
                new_distance = current_distance + weight

                if new_distance < distances[neighbor]:
                    distances[neighbor] = new_distance
                    previous[neighbor] = current_city

                    heapq.heappush(
                        queue, (new_distance, neighbor)
                    )

        if distances[destination] == float('inf'):
            print("No route available!")
            return

        path = []
        city = destination

        while city != start:
            path.append(city)
            city = previous[city]

        path.append(start)
        path.reverse()

        print("\nShortest Route:", " -> ".join(path))
        print("Total Distance:", distances[destination], "km")


gps = GPSNavigation()

# Adding roads and distances
gps.add_road("Pune", "Mumbai", 150)
gps.add_road("Pune", "Nashik", 210)
gps.add_road("Mumbai", "Nashik", 170)
gps.add_road("Mumbai", "Satara", 110)
gps.add_road("Satara", "Pune", 110)

print("===== GPS Navigation System =====")
start = input("Enter starting location: ").title()
destination = input("Enter destination: ").title()

if start not in gps.graph or destination not in gps.graph:
    print("Location not found!")
else:
    gps.shortest_path(start, destination)
```
